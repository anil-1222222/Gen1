GROQ MODEL Integration

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

# Use a currently active Groq model
model = init_chat_model("groq:llama-3.3-70b-versatile")

# Stream the response
for chunk in model.stream("what is java in 30 words"):
    print(chunk.content, end="", flush=True)
```

```python
import os
from langchain.chat_models import init_chat_model

# 1. Set environment key
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

# 2. Re-initialize model with an active model ID
model = init_chat_model("groq:llama-3.3-70b-versatile")

# 3. Process batch requests
responses = model.batch([
    "what is datatypes in java?",
    "what is object in java?",
    "What is class in java?"
])

# 4. Print response contents
for response in responses:
    print(response.content)
    print("=" * 60)
```

```python
responses = model.batch([
    "what is datatypes in java?",
    "what is object in java?",
    "What is class in java?"],
    config={
        'max_concurrency': 5,  # Limit to 5 parallel calls
    }
)

# 4. Print response contents
for response in responses:
    print(response.content)
    print("=" * 60)
```
