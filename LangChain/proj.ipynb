# Groq LangChain Model Streaming & Chrome Web UI
This notebook demonstrates real-time streaming using LangChain with Groq (`llama-3.3-70b-versatile`) and launching the interactive web interface.

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# Load environment variables
load_dotenv()
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

# Initialize Groq model
model = init_chat_model("groq:llama-3.3-70b-versatile")

# Stream response token by token in console
print("--- Real-time Response Stream ---\n")
for chunk in model.stream("what is java in 30 words"):
    print(chunk.content, end="", flush=True)
print("\n\n--- Stream Completed ---")
```

```python
# Launch Web Application UI in Chrome
import webbrowser

url = "http://localhost:5000"
print(f"Opening Chrome browser at: {url}")
webbrowser.open(url)
```
